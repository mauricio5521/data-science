### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

```python
# 1. Agregar una columna para identificar cada tienda antes de juntarlas
tienda['Nombre_Tienda'] = 'Tienda 1'
tienda2['Nombre_Tienda'] = 'Tienda 2'
tienda3['Nombre_Tienda'] = 'Tienda 3'
tienda4['Nombre_Tienda'] = 'Tienda 4'

# 2. Unificar (Concatenar) los 4 dataframes en uno solo
df_total = pd.concat([tienda, tienda2, tienda3, tienda4])

# 3. Reiniciar el índice para evitar duplicados en el índice
df_total.reset_index(drop=True, inplace=True)

# Verificamos cómo quedó la tabla maestra
print(f"Total de filas cargadas: {df_total.shape[0]}")
df_total.tail(50)
```

```python
df_total.info()
```

#1. Análisis de facturación



```python
facturacion = (
    df_total
    .groupby('Nombre_Tienda')['Precio']
    .sum()
    .sort_values(ascending=False)
)

print("FACTURACIÓN TOTAL POR TIENDA")
print(facturacion)
```

# 2. Ventas por categoría

```python
ventas_categoria = (
    df_total
    .groupby(['Nombre_Tienda', 'Categoría del Producto'])
    .size()
    .reset_index(name='Cantidad_Ventas')
    .sort_values(['Nombre_Tienda', 'Cantidad_Ventas'], ascending=[True, False])
)

print("\nVENTAS POR CATEGORÍA")
print(ventas_categoria)
```

# 3. Calificación promedio de la tienda


```python
calificacion_promedio = (
    df_total
    .groupby('Nombre_Tienda')['Calificación']
    .mean()
    .sort_values(ascending=False)
)

print("\nCALIFICACIÓN PROMEDIO POR TIENDA")
print(calificacion_promedio)
```

# 4. Productos más y menos vendidos

```python
productos_vendidos = (
    df_total
    .groupby(['Nombre_Tienda', 'Producto'])
    .size()
    .reset_index(name='Cantidad')
)

mas_vendidos = (
    productos_vendidos
    .sort_values(['Nombre_Tienda', 'Cantidad'], ascending=[True, False])
    .groupby('Nombre_Tienda')
    .first()
)

menos_vendidos = (
    productos_vendidos
    .sort_values(['Nombre_Tienda', 'Cantidad'], ascending=[True, True])
    .groupby('Nombre_Tienda')
    .first()
)

print("\nPRODUCTOS MÁS VENDIDOS POR TIENDA")
print(mas_vendidos)

print("\nPRODUCTOS MENOS VENDIDOS POR TIENDA")
print(menos_vendidos)


```

# 5. Envío promedio por tienda

```python
envio_promedio = (
    df_total
    .groupby('Nombre_Tienda')['Costo de envío']
    .mean()
    .sort_values()
)

print("\nCOSTO PROMEDIO DE ENVÍO POR TIENDA")
print(envio_promedio)

```

#**Decicsión final**

```python
resumen = pd.DataFrame({
    'Facturacion': facturacion,
    'Calificacion_Promedio': calificacion_promedio,
    'Envio_Promedio': envio_promedio
})

print("\nRESUMEN GENERAL")
print(resumen)

tienda_a_vender = resumen['Facturacion'].idxmin()

print(f"\nRECOMENDACIÓN FINAL:")
print(f"La tienda que presenta el menor desempeño general y debería considerarse para la venta es: {tienda_a_vender}")
```
